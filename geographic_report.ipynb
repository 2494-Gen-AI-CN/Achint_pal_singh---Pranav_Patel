## Dataset Overview

We filtered the original Dataset by the year 2024-2025, which contains 137,727 collision records. That dataset contains OCC_YEAR (the year of collision), NEIGHBOURHOOD_158 (the neighbourhood were collision was recorded), BICYCLE (if bicycle was involved), PEDESTRIAN (if pedestrian were involved), and lastly INJURY_COLLISIONS (whether injured occured during the collision)
